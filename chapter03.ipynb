## Defining New Functions

- 함수 정의(function definition): 새로운 함수의 이름과 해당 함수가 호출되었을 때 실행할 명령문들의 순서를 지정

````python
# 벌목꾼의 노래 가사 중 첫 두 줄을 출력하는 함수를 정의한다.
# 함수를 호출하면 두 개의 print 문이 순서대로 실행된다.


def print_lyrics():
    print("I'm a lumberjack, and I'm okay.")
    print("I sleep all night and I work all day.")
````

- 헤더(header): 함수 정의의 첫 줄
	- `def`: 함수 정의를 나타내는 예약어
	- `print_lyrics`: 위에서 정의한 함수의 이름
	- `()`: 이 함수가 인자를 받지 않음을 의미
	- 반드시 콜론(`:`)으로 끝남
- 본문(body): 나머지 부분
	- 반드시 들여쓰기
	- 위의 함수는 2개의 `print`문으로 구성
- 함수를 정의하면 함수 객체(function object)가 생성


````python
# 앞에서 정의한 print_lyrics 함수의 객체를 확인한다.
# 함수를 호출하지 않고 이름만 입력하여 함수 자체를 표시한다.


print_lyrics
````

````python
# 앞에서 정의한 print_lyrics 함수를 호출한다.
# 함수 본문에 작성된 두 개의 print 문이 실행되어 노래 가사가 출력된다.

print_lyrics()
````

## Parameters

````python
# 전달받은 문자열을 두 번 출력하는 함수를 정의한다.
# 매개변수 string에 전달된 값을 사용하여
# 두 개의 print 문을 순서대로 실행한다.


def print_twice(string):
    print(string)
    print(string)
````

- 매개변수(parameter): 함수 이름 뒤의 괄호 안에 있는 변수 이름
	- 함수가 호출되면 전달된 인자의 값이 매개변수에 할당

````python
# 문자열을 인자로 전달하여 print_twice 함수를 호출한다.
# 전달된 문자열이 매개변수 string에 할당되고
# 동일한 문자열이 두 번 출력된다.

print_twice("Dennis Moore, ")
````

````python
# 문자열을 변수 string에 할당한다.
# 이후 두 개의 print 문을 실행하여
# 변수에 저장된 문자열을 두 번 출력한다.

string = "Dennis Moore, "
print(string)
print(string)
````

````python
# 문자열을 변수 line에 저장한다.
# line을 인자로 전달하여 print_twice 함수를 호출하면
# 변수에 저장된 문자열이 두 번 출력된다.

line = "Dennis Moore, "
print_twice(line)
````

## Calling Functions

- 함수를 정의하면, 다른 함수 안에서도 해당 함수 사용 가능

````python
# 전달받은 단어나 문자열을 지정한 횟수만큼 반복하여 출력하는 함수이다.
# word에는 반복할 문자열이, n에는 반복 횟수가 전달된다.


def repeat(word, n):
    print(word * n)
````

````python
# spam 변수에 문자열을 저장한다.
# repeat 함수를 호출하여 해당 문자열을 네 번 반복해서 출력한다.

spam = "Spam, "
repeat(spam, 4)
````

````python
# 노래의 첫 두 줄을 출력하는 함수를 정의한다.
# repeat 함수를 두 번 호출하여
# 각 줄마다 Spam 문자열을 네 번씩 반복해서 출력한다.


def first_two_lines():
    repeat(spam, 4)
    repeat(spam, 4)
````

````python
# 앞에서 정의한 first_two_lines 함수를 호출한다.
# 함수 내부에서 repeat 함수가 두 번 실행되어
# 노래의 첫 두 줄이 출력된다.

first_two_lines()
````

````python
# 노래의 마지막 세 줄을 출력하는 함수를 정의한다.
# repeat 함수로 첫 번째 줄과 세 번째 줄을 출력하고
# 가운데 줄은 print 함수를 사용하여 출력한다.


def last_three_lines():
    repeat(spam, 2)
    print("(Lovely Spam, Wonderful Spam!)")
    repeat(spam, 2)
````

````python
# 앞에서 정의한 last_three_lines 함수를 호출한다.
# 함수 본문에 작성된 명령문들이 순서대로 실행되어
# 노래의 마지막 세 줄이 출력된다.

last_three_lines()
````

````python
# 앞에서 정의한 두 함수를 하나로 묶는 함수를 정의한다.
# first_two_lines와 last_three_lines를 순서대로 호출하여
# 노래의 한 절 전체를 출력한다.


def print_verse():
    first_two_lines()
    last_three_lines()
````

````python
# 앞에서 정의한 print_verse 함수를 호출한다.
# 함수 내부에서 다른 함수들이 차례대로 호출되며
# 노래의 한 절 전체가 출력된다.

print_verse()
````

## Repetition

````python
# range(2)를 사용하여 0과 1로 이루어진 값의 순서를 생성한다.
# for 반복문을 통해 각 값을 변수 i에 차례대로 할당하고 출력한다.

for i in range(2):
    print(i)
````

- 첫 번째 줄은 콜론(`:`)으로 끝나는 헤더
	- 예약어 `for`로 시작
	- 위의 예제는 `range` 함수를 사용하여 0과 1, 두 개의 값으로 이루어진 순서를 생성
- 두 번째 줄은 본문
	- 반드시 들여쓰기
- 반복문(repetition): 같은 작업을 반복하는 명령문
	- `for` 문이 실행되면 먼저 `range`에서 생성된 첫 번째 값(0)이 변수 `i`에 할당
	- 본문에 있는 `print` 함수가 실행되어 0이 출력
	- 본문의 실행이 끝나면 다시 헤더
	- 두 번째로 실행될 때는 `range`에서 생성된 다음 값(1)이 변수 `i`에 할당되고, 그 값을 출력
	- 이 값(1)이 `range`에서 생성된 마지막 값이므로 반복문은 종료

````python
# for 반복문을 사용하여 노래의 한 절을 두 번 출력한다.
# 각 반복마다 절 번호와 노래 가사를 출력하고,
# 마지막에 빈 줄을 추가하여 각 절을 구분한다.

for i in range(2):
    print("Verse", i)
    print_verse()
    print()
````

- `for` 반복문은 함수 내부에서도 사용 가능

````python
# 매개변수 n으로 전달받은 횟수만큼 노래의 한 절을 출력한다.
# for 반복문 안에서 print_verse 함수를 반복 호출하고,
# 각 절의 출력이 끝날 때마다 빈 줄을 추가한다.


def print_n_verses(n):
    for i in range(n):
        print_verse()
        print()
````

- 위의 예제에서 변수 `i`를 사용하지 않지만, `for`문의 헤더에는 반드시 변수 이름이 있어야 함

## Variables and Parameters Are Local

- 함수 내부에서 변수를 생성하면 그 변수는 지역 변수(local variable)
	- 지역 변수란 해당 변수가 함수 내부에서만 존재한다는 것을 의미

````python
# 두 개의 문자열을 인자로 받아 하나의 문자열로 연결하는 함수이다.
# 연결한 결과를 지역 변수 cat에 저장한 다음,
# print_twice 함수를 호출하여 결과를 두 번 출력한다.


def cat_twice(part1, part2):
    cat = part1 + part2
    print_twice(cat)
````

````python
# 두 개의 문자열을 각각 line1과 line2 변수에 저장한다.
# cat_twice 함수에 두 변수를 인자로 전달하여
# 연결된 문자열을 두 번 출력한다.

line1 = "Always look on the "
line2 = "bright side of life."
cat_twice(line1, line2)
````

- `cat_twice` 함수가 실행되면 `cat`이라는 지역 변수가 생성
	- `cat`은 함수의 실행이 종료되면 사라짐
	- 함수 외부에서 `cat`을 사용하면 `NameError` 발생

````python
# 함수 내부에서 생성된 지역 변수 cat에 접근하려는 코드이다.
# cat은 함수 외부에 정의되어 있지 않으므로
# NameError가 발생한다.

print(cat)
````

- 매개변수 역시 지역적
	- `cat_twice` 함수 외부에는 `part1`이나 `part2`라는 변수가 존재하지 않음

## Stack Diagrams

- 스택 다이어그램(stack diagram): 어떤 변수를 어디에서 사용할 수 있는지 파악하기 위한 다이어그램
	- 상태 다이어그램과 마찬가지로 변수의 값을 표현
	- 추가로 각 변수가 어떤 변수에 속하는지도 표현
	- 각각의 함수를 프레임(frame)으로 표현

- `__main__`
	- ```
	  line1 -> "Always look on the "
	  line2 -> "bright side of life."
	```
- `cat_twice`
	- ```
	  part1 -> "Always look on the "
	  part2 -> "bright side of life."
	  cat -> "Always look on the bright side of life."
	```
- `print_twice`
	- ```
	  string -> "Always look on the bright side of life."
	```
- `print`
	- ```
	  ? -> "Always look on the bright side of life."
	```

## Tracebacks

- 함수 내부에서 실행 시간 오류가 발생하면, 스택을 따라 올라가면서 함수 호출 관계를 표시
	- 오류가 발생했을 때 실행 중이던 함수의 이름
	- 위의 함수를 호출한 함수의 이름
	- 반복...

````python
# print_twice 함수를 오류가 발생하도록 새롭게 정의한다.
# 다른 함수의 지역 변수인 cat에 접근하려고 시도하므로
# NameError가 발생한다.


def print_twice(string):
    print(cat)  # NameError
    print(cat)
````

````python
# 앞에서 정의한 cat_twice 함수를 호출한다.
# 함수 내부에서 print_twice가 호출되지만,
# print_twice에서 정의되지 않은 cat에 접근하여 오류가 발생한다.

cat_twice(line1, line2)
````

- 트레이스백은 오류가 발생했을 때 실행 중이던 함수와 그 함수를 호출한 함수, 그리고 그 이전의 함수 호출 관계를 차례대로 출력
	- 표시되는 함수들의 순서는 스택 다이어그램과 동일
	- 오류가 발생한 함수는 트레이스백의 가장 아래에 표시

## Debugging

- 디버깅은 탐정의 사건 수사나 실험 과학과 비슷
	- 프로그램이 잘못된 이유에 대해 가설을 세우고, 프로그램을 수정하여 다시 실행
- 이미 정상적으로 작동하는 프로그램에서 시작하여 조금씩 수정하고, 수정할 때마다 디버깅
- 코드를 한 번에 많이 작성하기보다 작은 단계로 나누어 작성하고 테스트하면 더 빠르게 진행 가능
