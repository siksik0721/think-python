## Expressions

````python
# 거듭제곱과 덧셈을 함께 사용하는 표현식입니다.
# 거듭제곱을 먼저 계산한 다음 6을 더합니다.

6 + 6**2
````

````python
# 덧셈과 곱셈을 함께 사용하는 표현식입니다.
# 곱셈을 먼저 계산한 다음 12를 더합니다.

12 + 5 * 6
````

````python
# 괄호를 사용하여 덧셈을 먼저 수행하는 표현식입니다.
# 12와 5를 더한 다음 그 결과에 6을 곱합니다.

(12 + 5) * 6
````

표현식은 숫자와 연산자를 조합한 것으로, 계산하면 하나의 값을 가진다.

파이썬은 정해진 연산자 우선순위에 따라 표현식을 계산하며, 괄호를 사용하면 원하는 연산을 먼저 수행하도록 순서를 변경할 수 있다.

자연어는 사람들이 일상적으로 사용하는 언어이고, 형식 언어는 특정한 목적을 위해 설계된 언어이다. 프로그래밍 언어는 계산 과정을 명확하게 표현하기 위한 형식 언어에 해당된다.

## Debugging

디버깅은 프로그래밍에서 발생한 오류를 찾아내고 해결하는 과정이며, 프로그래밍을 배우는 데 반드시 필요한 중요한 기술이다.

디버깅 과정에서 좌절감이나 분노를 느낄 수 있지만, 컴퓨터의 장단점을 이해하고 자신의 감정을 적절히 조절하면서 문제를 해결하는 자세가 중요하다.

## Glossary

- 산술 연산자(arithmetic operator): 덧셈이나 곱셈과 같은 산술 연산을 나타내는 기호. 예를 들어 +는 덧셈을, *는 곱셈을 나타낸다.
- 정수(integer): 소수 부분이 없는 정수를 나타내는 자료형.
- 부동소수점(floating-point): 소수 부분이 있는 숫자를 나타내는 자료형.
- 정수 나눗셈(integer division): 두 숫자를 나눈 다음, 그 결과를 아래쪽 정수로 내림하는 연산. 파이썬에서는 // 연산자를 사용한다.
- 표현식(expression): 변수, 값, 연산자를 조합하여 만든 것.
- 값(value): 정수, 부동소수점 숫자, 문자열 또는 앞으로 살펴볼 다른 종류의 데이터.
- 함수(function): 특정한 유용한 작업을 수행하는, 이름이 붙은 명령문들의 나열. 함수는 인자를 받을 수도 있고 받지 않을 수도 있으며, 실행 결과를 생성할 수도 있고 생성하지 않을 수도 있다.
- 함수 호출(fuction call): 함수를 실행하는 표현식 또는 표현식의 일부분. 함수 호출은 함수 이름 뒤에 괄호로 감싼 인자 목록이 오는 형태로 구성된다.
- 구문 오류(syntax error): 프로그램의 문법이나 구조에 문제가 있어 코드를 구문 분석할 수 없고, 그 결과 프로그램을 실행할 수도 없게 만드는 오류.
- 문자열(string): 문자들이 순서대로 나열된 데이터를 나타내는 자료형.
- 문자열 연결(concatenation): 두 문자열을 끝과 끝이 이어지도록 결합하는 연산.
- 자료형(type): 값을 종류에 따라 구분한 범주. 지금까지 살펴본 자료형은 정수형 int, 부동소수점형 float, 문자열형 str이 있다.
- 피연산자(operand): 연산자가 연산을 수행하는 대상이 되는 값 중 하나.
- 자연어(natural language): 사람들이 일상적으로 사용하는 언어 중 자연스럽게 발전해 온 언어.
- 형식 언어(formal language): 수학적 개념이나 컴퓨터 프로그램을 표현하는 것처럼, 특정한 목적을 위해 사람들이 의도적으로 설계한 언어. 모든 프로그래밍 언어는 형식 언어에 해당한다.
- 버그(bug): 프로그램에서 발생하는 오류.
- 디버깅(debugging): 프로그램에서 발생한 오류를 찾아내고 수정하는 과정.

파이썬의 `round()` 함수는 숫자가 정확히 `.5`로 끝나는 경우 가장 가까운 짝수로 반올림합니다.
이를 은행가 반올림(Banker's Rounding) 또는 Round Half to Even이라고 합니다.

### 예시
```python
round(0.5)   # 0
round(1.5)   # 2
round(2.5)   # 2
round(3.5)   # 4
round(4.5)   # 4
round(5.5)   # 6
```

즉, 무조건 올림하거나 내림하는 것이 아니라 결과가 짝수가 되는 방향으로 반올림합니다.

### 왜 이런 방식을 사용할까요?
일반적인 사사오입 방식은 `.5`를 항상 올림하므로, 많은 숫자를 반올림하면 결과가 위쪽으로 치우칠 수 있습니다.
반면 은행가 반올림은 이러한 통계적 편향을 줄이기 위한 방식입니다.

### 일반적인 사사오입을 원한다면?
`decimal` 모듈의 `ROUND_HALF_UP`을 사용하면 됩니다.
```python
from decimal import Decimal, ROUND_HALF_UP

def round_half_up(x):
	return int(Decimal(str(x)).quantize(
		Decimal('1'),
		rounding=ROUND_HALF_UP
	))

round_half_up(0.5)  # 1
round_half_up(1.5)  # 2
round_half_up(2.5)  # 3
round_half_up(3.5)  # 4
```

참고: `round(2.675, 2)`처럼 소수점 자릿수를 지정할 때는 부동소수점 표현 오차 때문에 예상과 다른 결과가 나올 수도 있습니다.
