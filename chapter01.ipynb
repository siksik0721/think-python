## Arithmetic Operators

- 산술 연산자(arithmetic operator): +, - 등 산술 계산을 나타내는 기호

```python
# 두 정수 30과 12를 더하는 코드입니다.
# + 연산자는 덧셈을 수행합니다.

30 + 12
```

```python
# 정수 43에서 1을 빼는 코드입니다.
# - 연산자는 뺄셈을 수행합니다.

43 - 1
```

```python
# 정수 6과 7을 곱하는 코드입니다.
# * 연산자는 곱셈을 수행합니다.

6 * 7
```

```python
# 84를 2로 나누는 코드입니다.
# / 연산자를 사용한 나눗셈의 결과는 부동소수점 수가 됩니다.

84 / 2
```

- 두 정수를 더하거나(`+`), 빼거나(`-`), 곱하면(`*`) 정수(integer)
- 두 정수를 나누면(`/`) 부동소수점 수(float)
	- 정수 나눗셈(`//`)은 결과를 항상 작은 정수 방향으로 내림

```python
# 84를 2로 정수 나눗셈하는 코드입니다.
# // 연산자의 결과는 항상 정수입니다.

84 // 2
```

```python
# 85를 2로 정수 나눗셈하는 코드입니다.
# 실제 몫 42.5를 아래쪽 정수인 42로 내림합니다.

85 // 2
```

- `**` 연산자는 거듭제곱을 수행
	- 일부 다른 프로그래밍 언어는 `^` 연산자로 거듭제곱을 수행
	- 파이썬에서 `^`은 배타적 논리합(XOR)을 수행

```python
# 7을 2제곱하는 코드입니다.
# ** 연산자는 거듭제곱을 수행합니다.

7**2
```

```python
# 7과 2에 배타적 논리합 비트 연산을 수행하는 코드입니다.
# 파이썬에서 ^는 거듭제곱 연산자가 아닙니다.

7 ^ 2
```

## Expressions

- 표현식(expression): 연산자와 숫자를 모아 놓은 것

```python
# 거듭제곱과 덧셈을 함께 수행하는 표현식입니다.
# 거듭제곱이 먼저 계산된 다음 덧셈이 수행됩니다.

6 + 6**2
```

- 연산자 간에는 연산 순서가 존재
	- 거듭제곱 > 곱셈, 나눗셈 > 덧셈, 뺄셈

```python
# 덧셈과 곱셈을 함께 수행하는 표현식입니다.
# 연산 순서에 따라 곱셈이 덧셈보다 먼저 수행됩니다.

12 + 5 * 6
```

- 괄호를 사용하여 연산 순서를 바꿀 수 있음

```python
# 괄호를 사용하여 덧셈을 먼저 수행하는 표현식입니다.
# 괄호 안의 12 + 5를 계산한 다음 그 결과에 6을 곱합니다.

(12 + 5) * 6
```

- 모든 표현식은 값이 존재
	- ex> `6 * 7`의 값은 `42`

## Arithmetic Functions

- `round`: 부동소수점 수를 받아 가장 가까운 정수로 반올림

```python
# 부동소수점 수 42.4를 가장 가까운 정수로 반올림합니다.
# 42.4는 42에 더 가까우므로 결과는 42입니다.

round(42.4)
```

```python
# 부동소수점 수 42.6을 가장 가까운 정수로 반올림합니다.
# 42.6은 43에 더 가까우므로 결과는 43입니다.

round(42.6)
```

- `abs`: 절대값을 계산

```python
# 양수 42의 절댓값을 계산합니다.
# 양수의 절댓값은 그 수 자체이므로 결과는 42입니다.

abs(42)
```

```python
# 음수 -42의 절댓값을 계산합니다.
# 절댓값은 음수가 될 수 없으므로 결과는 양수 42입니다.

abs(-42)
```

- 함수 호출(function call): 함수를 사용하는 표현식
	- 반드시 괄호가 필요

```python
# abs 함수를 호출하면서 필요한 괄호를 생략한 잘못된 코드입니다.
# 함수 호출에는 괄호가 필요하므로 문법 오류가 발생합니다.

abs 42
```

- 위의 에러 메시지에서
	- 두 번째 줄에 오류가 발생한 코드와 위치를 출력
	- 마지막 줄에 어떤 오류가 발생했는지 출력
		- 문법 오류(syntax error): 표현식의 구조에 대한 오류

```python
# abs라는 함수 이름 자체를 하나의 표현식으로 사용합니다.
# 함수 이름만 작성하는 것은 올바른 표현식이므로 오류가 발생하지 않습니다.

abs
```

## Strings

- 문자열(string): 문자들이 연속해서 나열된 것
	- 파이썬에서는 따옴표 안에 문자들을 넣어 표현

```python
# 작은따옴표를 사용하여 문자열을 만드는 코드입니다.
# 따옴표 안의 문자들이 하나의 문자열을 구성합니다.

'Hello'
```

```python
# 큰따옴표를 사용하여 문자열을 만드는 코드입니다.
# 작은따옴표와 큰따옴표 모두 문자열을 표현하는 데 사용할 수 있습니다.

"world"
```

```python
# 큰따옴표를 사용하여 아포스트로피가 포함된 문자열을 만듭니다.
# 문자열 내부의 작은따옴표를 그대로 사용할 수 있습니다.

"it's a small"
```

- 문자열 안에 공백, 문장 부호, 숫자도 포함 가능

```python
# 문자와 문장 부호, 공백이 포함된 문자열을 만듭니다.
# 따옴표 안의 공백도 문자열의 일부로 취급됩니다.

"Well, "
```

- `+` 연산자는 두 문자열을 하나로 이어 붙이는 연산(concatenation)

```python
# + 연산자를 사용하여 세 문자열을 하나로 이어 붙입니다.
# 문자열 사이의 공백도 그대로 유지되어 하나의 문자열이 됩니다.

"Well, " + "it's a small " + 'world'
```

- `*` 연산자는 문자열을 복사해서 하나로 이어 붙이는 연산

```python
# 문자열을 네 번 복사한 뒤 하나로 이어 붙입니다.
# 문자열에 정수를 곱하면 해당 문자열을 지정한 횟수만큼 반복합니다.

"Spam, " * 4
```

- `+`, `*` 외의 산술 연산자는 문자열에 사용 불가
- `len`: 문자열의 길이를 계산
	- 따옴표 사이의 공백을 포함한 문자의 개수만 세고, 따옴표는 계산에 포함되지 않음

```python
# "Spam" 문자열의 길이를 계산합니다.
# len 함수는 문자열에 포함된 문자의 개수를 반환합니다.

len("Spam")
```

- 문자열은 반드시 따옴표(`''`, `""`) 사용
	- 백쿼트(backquote, ``)나 스마트 따옴표는 사용 불가

```python
# 백쿼트를 사용하여 문자열을 만들려고 한 잘못된 코드입니다.
# 파이썬에서는 백쿼트를 문자열의 따옴표로 사용할 수 없어 문법 오류가 발생합니다.

`Hello`
```

## Values and Types

- 자료형(type): 값(value)의 종류
	- ex> 정수(`int`), 부동소수점 수(`float`), 문자열(`str`)
- `type`: 어떤 값의 자료형을 알려줌

```python
# 정수 2의 자료형을 확인합니다.
# type 함수는 주어진 값이 어떤 자료형인지 알려줍니다.

type(2)
```

```python
# 부동소수점 수 42.0의 자료형을 확인합니다.
# 부동소수점 수의 자료형은 float입니다.

type(4.2)
```

```python
# 문자열의 자료형을 확인합니다.
# 문자열의 자료형은 str입니다.

type("Hello, World!")
```

- `int`, `float`, `str` 자료형은 함수처럼 사용하면 자료형 변환을 수행

```python
# 부동소수점 수 42.9를 정수로 변환합니다.
# int를 사용하면 소수점 이하 부분이 제거되어 42가 됩니다.

int(42.9)
```

```python
# 정수 42를 부동소수점 수로 변환합니다.
# float를 사용하면 42가 42.0으로 변환됩니다.

float(42)
```

```python
# 숫자처럼 보이는 문자들을 따옴표 안에 넣어 문자열을 만듭니다.
# 따옴표 안에 있기 때문에 숫자가 아니라 문자열로 취급됩니다.

"126"
```

```python
# '126'의 자료형을 확인합니다.
# 숫자로만 구성되어 있어도 따옴표 안에 있으면 str 자료형입니다.

type("126")
```

```python
# 문자열 '126'을 정수 3으로 나누려고 합니다.
# 문자열과 정수에는 / 연산을 적용할 수 없으므로 자료형 오류가 발생합니다.

"126" / 3
```

- 자료형 오류(type error): 표현식에 들어있는 값의 자료형이 잘못되었음

```python
# 문자열 '126'을 정수로 변환한 뒤 3으로 나눕니다.
# 정수끼리 / 연산을 수행하므로 결과는 부동소수점 수가 됩니다.

int("126") / 3
```

```python
# 문자열 '12.6'을 부동소수점 수로 변환합니다.
# 소수점이 포함된 숫자 문자열을 float 자료형으로 바꿀 수 있습니다.

float("12.6")
```

- 큰 정수를 사용할 때 자릿수를 구분하고 싶으면 `,`가 아닌 `_`을 사용

```python
# 숫자 사이를 쉼표로 구분한 표현식입니다.
# 파이썬은 이것을 하나의 큰 정수가 아니라 쉼표로 구분된 정수들의 나열로 해석합니다.

1,000,000
```

```python
# 밑줄을 사용하여 큰 정수를 읽기 쉽게 표현합니다.
# 밑줄은 숫자의 값에 영향을 주지 않으므로 결과는 정수 1000000입니다.

1_000_000
```

## Formal and Natural Languages

- 자연어(natural language): 영어, 스페인어처럼 사람들이 사용하는 언어
- 형식 언어(formal language): 특정한 용도를 위해 사람들이 설계한 언어
- 두 언어의 차이점
	- 모호성(ambiguity): 자연어는 모호한 표현이 많지만, 형식 언어는 정확히 하나의 의미를 가짐
	- 중복성(redundancy): 자연어는 오해를 줄이기 위해 중복된 표현을 사용하지만, 형식 언어는 중복이 적고 간결함
	- 문자 그대로의 의미(literalness): 자연어는 관용구나 비유가 많지만, 형식 언어는 작성된 내용 그대로 의미

## Debugging

- 디버깅(debugging): 오류를 찾아내는 과정
