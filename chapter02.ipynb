## Variables

```python
# 정수 17을 n이라는 변수에 대입합니다.
# 이후 n을 사용하여 저장된 값을 참조할 수 있습니다.

n = 17
```

```python
# 원주율의 근삿값을 pi라는 변수에 대입합니다.
# 이 값은 부동소수점 수입니다.

pi = 3.141592653589793
```

```python
# 문자열을 message라는 변수에 대입합니다.
# 이후 message를 사용하여 해당 문자열을 참조할 수 있습니다.

message = "And now for something completely different"
```

```python
# message 변수를 평가합니다.
# 변수에 저장된 문자열 값이 표시됩니다.

message
```

```python
# n에 저장된 값 17에 25를 더합니다.
# 계산 결과는 42입니다.

n + 25
```

```python
# pi에 저장된 원주율의 근삿값에 2를 곱합니다.
# 계산 결과는 부동소수점 수로 나타납니다.

2 * pi
```

```python
# round 함수에 pi 변수를 전달합니다.
# pi의 값을 가장 가까운 정수로 반올림합니다.

round(pi)
```

```python
# len 함수에 message 변수를 전달합니다.
# 문자열에 포함된 문자의 개수를 구합니다.

len(message)
```

변수는 값을 가리키는 이름이며, 대입 연산자(`=`)를 사용하여 생성할 수 있다. 대입문을 실행해도 화면에 결과가 출력되지 않는다.

생성한 변수는 독립적인 표현식으로 사용할 수 있으며, 산술 연산이나 함수 호출에도 활용할 수 있다.

## State Diagrams

상태 다이어그램은 변수의 이름과 그 변수가 가리키는 값을 화살표로 연결하여 나타낸 그림이다.

상태 다이어그램을 활용하면 파이썬이 변수와 값을 저장하는 방식을 시각적으로 설명할 수 있다.

## Variable Names

```python
# 변수 이름에 느낌표가 포함되어 있습니다.
# 느낌표는 변수 이름에 사용할 수 없으므로 구문 오류가 발생합니다.

million! = 1000000
```

```python
# 변수 이름이 숫자 76으로 시작합니다.
# 변수 이름은 숫자로 시작할 수 없으므로 구문 오류가 발생합니다.

76trombones = "big parade"
```

```python
# class를 변수 이름으로 사용하려고 합니다.
# class는 파이썬의 예약된 키워드이므로 구문 오류가 발생합니다.

class = "Self-Defence Against Fresh Fruit"
```

변수 이름에는 문자, 숫자, 밑줄(`_`)을 사용할 수 있지만 숫자로 시작할 수는 없으며, 일반적으로 소문자를 사용하는 것이 관례이다. 또한 파이썬에서 특별한 의미를 갖는 키워드는 변수 이름으로 사용할 수 없다.

올바르지 않은 변수 이름을 사용하면 구문 오류가 발생한다.

## The `import` Statement

```python
# 수학 계산에 필요한 기능을 사용하기 위해 math 모듈을 가져옵니다.
# 이 모듈에는 수학 상수와 다양한 수학 함수가 포함되어 있습니다.

import math
```

```python
# math 모듈에 포함된 pi 변수에 접근합니다.
# 이 변수에는 원주율의 근삿값이 저장되어 있습니다.

math.pi
```

```python
# math 모듈의 sqrt 함수를 사용합니다.
# 25의 제곱근을 계산합니다.

math.sqrt(25)
```

```python
# math 모듈의 pow 함수를 사용합니다.
# 5의 2제곱을 계산합니다.

math.pow(5, 2)
```

파이썬의 일부 기능을 사용하려면 `import` 문으로 모듈을 가져와야 한다. 모듈은 변수와 함수의 모음이며, 점 연산자(`.`)를 사용하여 모듈에 포함된 기능에 접근할 수 있다.

`math` 모듈은 원주율과 제곱근, 거듭제곱 등의 수학 기능을 제공한다. 거듭제곱은 `math.pow` 함수 또는 `**` 연산자로 계산할 수 있으며, 일반적으로 연산자를 더 자주 사용한다.

## Expressions and Statements

```python
# 변수 n의 값과 math.pi를 반올림한 값을 사용합니다.
# 덧셈과 곱셈을 수행하여 최종 결과를 계산합니다.

19 + n + round(math.pi) * 2
```

```python
# n이라는 변수를 생성하고 정수 17을 대입합니다.
# 대입문 자체는 값을 갖지 않습니다.

n = 17
```

```python
# math 모듈을 가져옵니다.
# 모듈에 포함된 변수와 함수를 사용할 수 있게 되지만, 화면에는 아무것도 출력되지 않습니다.

import math
```

표현식은 값, 연산자, 변수, 함수 호출 등으로 구성될 수 있으며, 평가를 통해 하나의 값을 얻는다. 문장은 대입이나 모듈 가져오기처럼 특정한 동작을 수행하지만, 문장 자체는 값을 갖지 않는다.

표현식은 평가하고, 문장은 실행한다는 것이 핵심적인 차이이다.

## The `print` Function

```python
# 변수 n에 저장된 값 17에 1을 더합니다.
# 표현식을 평가한 결과가 화면에 표시됩니다.

n + 1
```

```python
# 두 개의 표현식을 순서대로 평가합니다.
# 마지막 표현식인 n + 3의 결과만 화면에 표시됩니다.

n + 2
n + 3
```

```python
# print 함수를 사용하여 두 표현식의 결과를 각각 출력합니다.
# 각 함수 호출의 결과가 별도의 줄에 표시됩니다.

print(n + 2)
print(n + 3)
```

```python
# 첫 번째 print 함수는 문자열을 출력합니다.
# 두 번째 print 함수는 원주율의 근삿값을 출력합니다.

print("The value of pi is approximately")
print(math.pi)
```

```python
# 문자열과 math.pi 변수를 쉼표로 구분하여 전달합니다.
# 두 값을 한 줄에 출력하며, 값 사이에는 공백이 자동으로 삽입됩니다.

print("The value of pi is approximately", math.pi)
```

여러 표현식을 연속해서 평가하면 마지막 표현식의 결과만 표시되지만, `print` 함수를 사용하면 여러 값을 각각 출력할 수 있다.

`print` 함수에는 쉼표로 구분한 여러 표현식을 전달할 수 있으며, 이 때 출력되는 값 사이에는 공백이 자동으로 삽입된다.

## Arguments

```python
# int 함수에 문자열 "101"을 인자로 전달합니다.
# 문자열을 정수로 변환합니다.

int("101")
```

```python
# math.pow 함수에 두 개의 인자 5와 2를 전달합니다.
# 첫 번째 인자인 5를 두 번째 인자인 2만큼 거듭제곱합니다.

math.pow(5, 2)
```

```python
# int 함수에 문자열 "101"과 진법을 나타내는 2를 전달합니다.
# 이진수 101을 십진수 정수로 변환합니다.

int("101", 2)
```

```python
# round 함수에 원주율과 소수점 이하 자릿수 3을 전달합니다.
# 원주율을 소수점 이하 셋째 자리까지 반올림합니다.

round(math.pi, 3)
```

```python
# print 함수에 네 개의 문자열을 인자로 전달합니다.
# 각 문자열 사이에 공백을 넣어 한 줄로 출력합니다.

print("Any", "number", "of", "arguments")
```

```python
# float 함수에 두 개의 인자를 전달합니다.
# float 함수는 인자를 최대 하나만 받으므로 자료형 오류가 발생합니다.

float("123.0", 2)
```

```python
# math.pow 함수에 인자 하나만 전달합니다.
# 이 함수에는 두 개의 인자가 필요하므로 자료형 오류가 발생합니다.

math.pow(2)
```

```python
# math.sqrt 함수에 숫자가 아닌 문자열을 전달합니다.
# 이 함수는 문자열을 처리할 수 없으므로 자료형 오류가 발생합니다.

math.sqrt("123")
```

인자는 함수를 호출할 때 괄호 안에 전달하는 표현식이다. 함수마다 필요한 인자의 개수가 다르며, 일부 함수는 선택적인 추가 인자나 개수에 제한이 없는 인자를 받을 수 있다.

함수에 너무 많거나 적은 인자를 전달하거나, 처리할 수 없는 자료형을 전달하면 자료형 오류가 발생한다. 이러한 오류 검사는 프로그램의 문제를 발견하고 수정하는 데 도움을 준다.

## Comments

```python
# 42분 42초의 총 초 수

seconds = 42 * 60 + 42
```

```python
miles = 10 / 1.61  # 10킬로미터를 마일로 변환
```

```python
# 주석이 코드의 동작을 그대로 반복하므로 추가적인 정보를 제공하지 않습니다.

v = 8  # v에 8을 대입
```

```python
# 주석을 통해 이 값의 단위가 시간당 마일임을 알 수 있습니다.

v = 8  # 시간당 마일 단위의 속도
```

주석은 `#` 기호로 시작하며, 프로그램의 동작을 이해하는 데 도움을 주는 설명이다. 주석은 파이썬이 실행할 때 무시하므로 프로그램의 실행 결과에 영향을 주지 않는다.

좋은 주석은 코드의 동작을 단순히 반복하기보다 코드가 왜 그렇게 작성되었는지, 코드만으로 알기 어려운 정보가 무엇인지 설명한다. 또한 명확한 변수 이름과 주석을 적절히 활용하여 코드의 가독성을 높이는 것이 중요하다.

## Debugging

```python
# 변수 이름에 느낌표가 포함되어 있습니다.
# 느낌표는 변수 이름에 사용할 수 없으므로 구문 오류가 발생합니다.

million! = 1000000
```

```python
# 문자열 "126"을 정수 3으로 나누려고 합니다.
# 나눗셈 연산자는 문자열과 정수 사이의 연산을 지원하지 않으므로 실행 시간 오류가 발생합니다.

"126" / 3
```

```python
# 숫자 1과 3의 평균을 계산하려는 표현식입니다.
# 그러나 나눗셈이 덧셈보다 먼저 수행되므로 의도한 평균값이 나오지 않습니다.

1 + 3 / 2
```

프로그램의 오류는 구문 오류, 실행 시간 오류, 의미 오류로 구분할 수 있다. 구문 오류는 실행을 막고, 실행 시간 오류는 실행 도중 프로그램을 중단시킨다. 반면 의미 오류는 프로그램이 정상적으로 실행되더라도 잘못된 결과를 만들어 내는 오류이다. 따라서 오류 메시지가 없다고 해서 프로그램이 반드시 올바른 것은 아니다.

## Glossary

- 변수(variable): 값을 가리키는 이름.
- 대입문(assignment statement): 변수에 값을 대입하는 문장.
- 상태 다이어그램(state diagram): 여러 변수와 각 변수가 가리키는 값을 그림으로 나타낸 것.
- 키워드(keyword): 프로그램의 구조를 지정하는 데 사용되는 특별한 단어.
- 가져오기 문장(import statement): 모듈 파일을 읽어 들여 해당 모듈에 포함된 변수와 함수를 사용할 수 있도록 하는 문장.
- 모듈(module): 함수 정의와 때로는 다른 문장들을 포함하는 파이썬 코드가 저장된 파일.
- 점 연산자(dot operator): 다른 모듈에 포함된 함수에 접근할 때 사용하는 연산자(.). 모듈 이름 뒤에 점을 찍고 함수 이름을 작성하는 방식으로 사용한다.
- 문장(statement): 명령이나 동작을 나타내는 하나 이상의 코드 줄.
- 평가(evaluate): 표현식에 포함된 연산을 수행하여 값을 계산하는 과정.
- 실행(execute): 문장을 실행하여 해당 문장이 지시하는 작업을 수행하는 것.
- 인자(argument): 함수를 호출할 때 함수에 전달하는 값.
- 주석(comment): 프로그램에 대한 정보를 제공하기 위해 코드에 포함하는 설명문. 프로그램의 실행에는 영향을 미치지 않는다.
- 실행 시간 오류(runtime error): 프로그램이 오류 메시지를 표시하고 종료되도록 만드는 오류.
- 예외(exception): 프로그램이 실행되는 도중 감지되는 오류.
- 의미 오류(semantic error): 프로그램이 오류 메시지를 표시하지는 않지만, 의도한 것과 다른 작업을 수행하도록 만드는 오류.
