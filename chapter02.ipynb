# CHAPTER 2. Variables and Statements

## Variables

- 변수(variable): 어떤 값을 가리키는 이름
	- 대입문(assignment statement)을 사용하여 변수에 값을 넣음

```python
# 변수 n을 만들고 정수 값 17을 대입합니다.

n = 17
```

- 대입문은 세 부분으로 구성
	- 왼쪽은 변수의 이름
	- 가운데는 등호 연산자(`=`)
	- 오른쪽은 표현식

```python
# 변수 pi를 만들고 원주율을 나타내는 부동소수점 값을 대입합니다.

pi = 3.141592653589793
```

```python
# 변수 message를 만들고 문자열을 대입합니다.

message = "And now for something completely different"
```

- 대입문은 실행해도 출력되는 것은 없음
- 변수를 만든 뒤에는 그 변수를 표현식으로 사용 가능

```python
# 변수 message를 표현식으로 사용하여 저장된 값을 표시합니다.

message
```

```python
# 변수 n의 값에 25를 더합니다.

n + 25
```

```python
# 변수 pi의 값에 2를 곱합니다.

2 * pi
```

```python
# pi의 값을 가장 가까운 정수로 반올림합니다.

round(pi)
```

```python
# message에 저장된 문자열의 길이, 즉 문자 개수를 구합니다.

len(message)
```

## State Diagrams

- 상태 다이어그램(state diagram): 변수를 나타낼 때 사용하는 방법 중 하나
	- 변수의 이름
	- 해당 변수가 가리키는 값으로 화살표 연결

- n -> 17
- pi -> 3.141592653589793
- message -> "And now for something completely different"

## Variable Names

- 변수의 이름에는 길이 제한 없음
- 문자와 숫자 모두 사용 가능하지만, 숫자로 시작은 불가
- 관례적으로 소문자만 사용
- `_`는 사용 가능한 유일한 문장 부호
- 허용되지 않는 형태로 변수의 이름을 사용하면 구문 오류 발생

```python
# 문장 부호인 느낌표가 포함된 잘못된 변수 이름을 사용합니다.
# 파이썬에서는 이런 변수 이름을 사용할 수 없으므로 구문 오류가 발생합니다.

million! = 1000000
```

```python
# 숫자로 시작하는 잘못된 변수 이름을 사용합니다.
# 파이썬의 변수 이름은 숫자로 시작할 수 없으므로 오류가 발생합니다.

76trombones = "big parade"
```

```python
# class를 변수 이름으로 사용하려고 합니다.
# class는 파이썬에서 특별한 의미를 가진 예약어이므로 변수 이름으로 사용할 수 없습니다.

class = "Self-Defence Against Fresh Fruit"
```

- 키워드(keyword): 파이썬에서 프로그램의 구조를 지정하는 데 사용되는 특별한 단어
	- 키워드는 변수 이름으로 사용 불가

## The import Statement

- 임포트(import): 파이썬의 일부 기능을 사용하기 위해 해당 기능을 가져오는 것

```python
# 수학과 관련된 변수와 함수를 사용할 수 있도록 math 모듈을 가져옵니다.

import math
```

- 모듈(module): 변수와 함수의 모음
	- 위에서 가져온 `math`는 수학과 관련된 모듈

```python
# math 모듈에 들어 있는 pi 변수의 값을 표시합니다.

math.pi
```

- 모듈의 변수나 함수를 사용하려면 모듈 이름과 변수 및 함수 이름 사이에 점 연산자(`.`)를 사용
- `math.sqrt`: 제곱근을 계산

```python
# math 모듈의 sqrt 함수를 사용하여 25의 제곱근을 계산합니다.

math.sqrt(25)
```

- `math.pow`: 첫 번째 수를 두 번째 수만큼 거듭제곱

```python
# math 모듈의 pow 함수를 사용하여 5의 2제곱을 계산합니다.

math.pow(5, 2)
```

## Expressions and Statements

```python
# 변수 n, 함수 호출, 원주율 값, 산술 연산자를 함께 사용하여 하나의 값을 계산합니다.
# n이 17일 때 이 표현식의 계산 결과는 42입니다.

19 + n + round(math.pi) * 2
```

- 문장(statement): 어떤 효과를 만들어 내지만, 그 자체로는 값을 가지지 않는 코드 단위
	- ex-1> 대입문은 변수를 만들고 값을 부여하지만, 대입문 자체에는 값이 없음
	- ex-2> 임포트 문도 모듈을 가져오지만, 값을 가지지 않음

```python
# 변수 n을 만들고 정수 값 17을 대입합니다.
# 변수와 값을 만들어 내는 효과가 있지만, 이 문장 자체에는 값이 없습니다.

n = 17
```

```python
# math 모듈을 가져와 그 안의 변수와 함수를 사용할 수 있게 합니다.
# 가져오기 문을 실행해도 화면에 직접 표시되는 결과는 없습니다.

import math
```

- 평가(evaluation): 표현식의 값을 계산
- 실행(execution): 문장을 실행

## The `print` Function

```python
# 변수 n의 값에 1을 더한 표현식을 평가합니다.
# n이 17이므로 결과로 18이 표시됩니다.

n + 1
```

- 두 개 이상의 표현식을 평가하면 마지막 표현식의 값만 출력

```python
# 두 개의 표현식을 차례로 평가합니다.
# 마지막 표현식의 값만 화면에 표시됩니다.

n + 2
n + 3
```

- 두 개 이상의 값을 표시하려면 `print` 함수 사용

```python
# print 함수를 각각 호출하여 두 표현식의 값을 모두 화면에 표시합니다.

print(n + 2)
print(n + 3)
```

```python
# 첫 번째 print 함수는 문자열을 출력합니다.
# 두 번째 print 함수는 math.pi에 저장된 원주율 값을 출력합니다.

print("The value of pi is approximately")
print(math.pi)
```

- 쉼표로 구분된 여러 표현식을 한 번에 출력 가능
	- 이 경우, 각각의 값 사이에 공백을 하나 넣어 출력

```python
# 하나의 print 함수에 쉼표로 구분된 여러 표현식을 전달합니다.
# 문자열과 원주율의 값을 한 줄에 함께 출력합니다.

print("The value of pi is approximately", math.pi)
```

## Arguments

- 인자(argument): 함수를 호출할 때, 괄호 안에 들어가는 표현식

```python
# 문자열 '101'을 정수로 변환합니다.

int("101")
```

```python
# 첫 번째 인자 5를 두 번째 인자 2만큼 거듭제곱합니다.

math.pow(5, 2)
```

- 일부 함수는 선택적 인자를 추가로 받을 수 있음

```python
# 문자열 '101'을 2진수로 해석하여 정수로 변환합니다.
# 2진수 101은 10진수로 5입니다.

int("101", 2)
```

```python
# 원주율 값을 소수점 아래 세 자리까지 반올림합니다.

round(math.pi, 3)
```

```python
# print 함수에 여러 개의 문자열을 인자로 전달합니다.
# 각 인자는 공백으로 구분되어 한 줄에 출력됩니다.

print("Any", "number", "of", "arguments")
```

- 너무 많은/적은 인자 혹은 함수가 처리할 수 없는 자료형의 인자를 전달하면 자료형 오류 발생

```python
# float 함수가 받을 수 있는 것보다 많은 인자를 전달합니다.
# 따라서 자료형 오류가 발생합니다.

float("123.0", 2)
```

```python
# math.pow 함수에는 두 개의 인자가 필요하지만 하나만 전달합니다.
# 필요한 인자가 부족하므로 자료형 오류가 발생합니다.

math.pow(2)
```

```python
# math.sqrt 함수에 숫자가 아닌 문자열을 전달합니다.
# 이 함수가 처리할 수 없는 자료형이므로 자료형 오류가 발생합니다.

math.sqrt("123")
```

## Comments

- 주석(comment): 프로그램 코드에 대해 자연어로 설명하는 메모
	- `#` 기호로 시작
	- `#`부터 그 줄의 끝까지 모든 내용을 파이썬이 무시하므로 프로그램 실행에 영향을 주지 않음
	- 무엇을 하는지는 코드를 보고 알아낼 수 있으므로 왜 그렇게 하는지를 설명하는 것이 유용

```python
# number of seconds in 42:42
seconds = 42 * 60 + 42
```

```python
miles = 10 / 1.61  # 10 kilometers in miles
```

```python
# 변수 v에 8을 대입하는 코드이며, 기존 주석도 같은 내용을 반복할 뿐입니다.

v = 8  # assign 8 to v
```

```python
# 값 8이 시간당 마일 단위의 속도를 의미한다는 추가 정보를 제공합니다.

v = 8  # velocity in miles per hour
```

## Debugging

- 구문 오류(syntax error): 프로그램 구조와 규칙에 대한 오류
	- 구문 오류가 있으면 파이썬은 프로그램을 실행하지 않고 오류 메시지를 표시
- 실행시간 오류(runtime error): 실행 도중 잘못된 오류
	- 실행 중 발생한 오류 메시지를 표시하고 실행을 중단
- 의미 오류(semantic error): 프로그래머의 의도와 다르게 동작하는 오류
	- 오류 메시지 없이 정상적으로 실행되지만, 실행 결과가 의도와 달라짐
	- 프로그램의 출력 결과를 거꾸로 추적하며 직접 잘못된 부분을 찾아야 함

```python
# 허용되지 않는 문장 부호가 포함된 변수 이름을 사용합니다.
# 변수 이름 규칙에 맞지 않으므로 구문 오류가 발생합니다.

million! = 1000000
```

```python
# 문자열을 정수로 나누려고 합니다.
# 나눗셈 연산자가 이 자료형의 조합을 지원하지 않으므로 실행 시간 오류가 발생합니다.

"126" / 3
```

```python
# 1과 3의 평균을 구하려고 작성한 표현식입니다.
# 하지만 연산 순서 때문에 의도했던 평균과 다른 값이 계산됩니다.

1 + 3 / 2
```
